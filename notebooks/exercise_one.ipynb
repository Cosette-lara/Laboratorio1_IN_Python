```python
import pandas as pd
df = pd.read_csv("../data/data_wine-1-1.csv", sep=";") # Leer el CSV indicando que el separador es un ";"

df.columns = df.columns.str.strip() # Eliminar espacios desde el inicio y al final de los nombres de columnas

print(df.columns) # Verifica las columnas disponibles

filas, columnas = df.shape # Cantidad de filas y columnas
print("Cantidad de filas:", filas)
print("Cantidad de columnas:", columnas)

print("Media Magnesium:", df["Magnesium"].mean()) # Media de Magnesium

print("Mediana Alcohol:", df["Alcohol"].median()) # Mediana de Alcohol

print("Suma Magnesium:", df["Magnesium"].sum()) # Suma de Magnesium
```
